# MODULE 2: UNIVARIATE FORECASTING

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import warnings
warnings.filterwarnings('ignore')

from statsmodels.tsa.stattools import adfuller, kpss, acf, pacf
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
from statsmodels.tsa.arima.model import ARIMA
from statsmodels.tsa.statespace.sarimax import SARIMAX
from statsmodels.stats.diagnostic import acorr_ljungbox
from statsmodels.graphics.gofplots import qqplot

from sklearn.metrics import mean_squared_error, mean_absolute_error
import itertools

plt.rcParams['figure.figsize'] = (14, 5)
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3

# Targets to model
TARGETS = ['nifty50_close', 'sp500_close', 'gold_close', 'usd_inr_close']
TEST_MONTHS = 24   # hold-out size
FORECAST_HORIZON = 12  # months to forecast forward
````

````python
df_raw = pd.read_csv('monthly_macro_data_ffill_bfill.csv', parse_dates=['date'])
df_raw = df_raw.sort_values('date').reset_index(drop=True)
df_raw = df_raw.set_index('date')
df_raw.index.freq = 'MS'  # Month Start frequency

print(f"Shape: {df_raw.shape}")
print(f"Date range: {df_raw.index.min()} → {df_raw.index.max()}")
df_raw[TARGETS].head()
````

````python
# Evaluation metrics
def evaluate_forecast(actual, predicted, label=''):
    """Returns a dict of RMSE, MAE, MAPE for a forecast."""
    actual = np.array(actual)
    predicted = np.array(predicted)
    rmse = np.sqrt(mean_squared_error(actual, predicted))
    mae  = mean_absolute_error(actual, predicted)
    mape = np.mean(np.abs((actual - predicted) / actual)) * 100
    if label:
        print(f"  [{label}]  RMSE: {rmse:.4f}  |  MAE: {mae:.4f}  |  MAPE: {mape:.2f}%")
    return {'RMSE': rmse, 'MAE': mae, 'MAPE': mape}


# Stationarity tests
def stationarity_tests(series, name=''):
    """Run ADF and KPSS tests; print results."""
    adf_stat, adf_p, _, _, adf_cv, _ = adfuller(series.dropna())
    kpss_stat, kpss_p, _, kpss_cv = kpss(series.dropna(), regression='c', nlags='auto')

    adf_result  = 'STATIONARY' if adf_p < 0.05 else 'NON-STATIONARY'
    kpss_result = 'STATIONARY' if kpss_p > 0.05 else 'NON-STATIONARY'

    print(f"\n{'='*55}")
    print(f" Stationarity Tests — {name}")
    print(f"{'='*55}")
    print(f"  ADF  stat={adf_stat:.4f}  p={adf_p:.4f}  → {adf_result}")
    print(f"  KPSS stat={kpss_stat:.4f}  p={kpss_p:.4f}  → {kpss_result}")

    return {'adf_p': adf_p, 'kpss_p': kpss_p,
            'adf_stationary': adf_p < 0.05,
            'kpss_stationary': kpss_p > 0.05}


# Plot ACF / PACF
def plot_acf_pacf(series, name='', lags=24):
    fig, axes = plt.subplots(1, 2, figsize=(14, 4))
    plot_acf(series.dropna(),  ax=axes[0], lags=lags, title=f'ACF  — {name}')
    plot_pacf(series.dropna(), ax=axes[1], lags=lags, title=f'PACF — {name}')
    plt.tight_layout()
    plt.show()


# Grid search ARIMA orders
def arima_grid_search(train, p_range=(0,3), d_range=(0,2), q_range=(0,3), verbose=False):
    """Search best (p,d,q) by AIC. Returns sorted results DataFrame."""
    results = []
    orders  = list(itertools.product(range(*p_range),
                                     range(*d_range),
                                     range(*q_range)))
    for order in orders:
        try:
            m = ARIMA(train, order=order).fit()
            results.append({'order': order, 'AIC': m.aic, 'BIC': m.bic})
            if verbose:
                print(f"  ARIMA{order}  AIC={m.aic:.2f}  BIC={m.bic:.2f}")
        except Exception:
            pass
    results_df = pd.DataFrame(results).sort_values('AIC').reset_index(drop=True)
    return results_df


# Residual diagnostics
def residual_diagnostics(model_fit, name=''):
    """Plot residual diagnostics and run Ljung-Box test."""
    resid = model_fit.resid.dropna()

    fig, axes = plt.subplots(2, 2, figsize=(14, 8))
    fig.suptitle(f'Residual Diagnostics — {name}', fontsize=13, fontweight='bold')

    # Residuals over time
    axes[0,0].plot(resid)
    axes[0,0].axhline(0, color='red', linestyle='--')
    axes[0,0].set_title('Residuals Over Time')

    # Histogram
    axes[0,1].hist(resid, bins=30, edgecolor='black')
    axes[0,1].set_title('Residuals Histogram')

    # ACF of residuals
    plot_acf(resid, ax=axes[1,0], lags=20, title='ACF of Residuals')

    # QQ plot
    qqplot(resid, line='s', ax=axes[1,1])
    axes[1,1].set_title('Q-Q Plot')

    plt.tight_layout()
    plt.show()

    # Ljung-Box test
    lb = acorr_ljungbox(resid, lags=[10, 20], return_df=True)
    print(f"\nLjung-Box Test ({name}):")
    print(lb)
    if (lb['lb_pvalue'] > 0.05).all():
        print("  ✓ Residuals look like white noise (no autocorrelation).")
    else:
        print("  ✗ Residuals show autocorrelation — model may be mis-specified.")


# Plot forecast vs actual
def plot_forecast(train, test, forecast_mean, conf_int=None, name='', model_label='ARIMA'):
    fig, ax = plt.subplots(figsize=(15, 5))

    # Show last 48 months of train for context
    ax.plot(train.index[-48:], train.values[-48:], label='Train (last 48m)', color='steelblue')
    ax.plot(test.index, test.values, label='Actual (Test)', color='black', linewidth=2)
    ax.plot(test.index, forecast_mean, label=f'{model_label} Forecast', color='crimson',
            linestyle='--', linewidth=2)

    if conf_int is not None:
        ax.fill_between(test.index,
                        conf_int.iloc[:, 0],
                        conf_int.iloc[:, 1],
                        alpha=0.2, color='crimson', label='95% CI')

    ax.set_title(f'{name} — {model_label} Forecast vs Actual', fontsize=13)
    ax.legend()
    plt.tight_layout()
    plt.show()
````

````python
# Stationarity analysis
stationarity_results = {}

for col in TARGETS:
    series = df_raw[col].dropna()
    print(f"\n{'─'*55}")
    print(f" {col.upper()}")

    # Level
    res_level = stationarity_tests(series, name=f'{col} (level)')

    # First difference
    res_diff  = stationarity_tests(series.diff().dropna(), name=f'{col} (1st diff)')

    stationarity_results[col] = {'level': res_level, 'diff': res_diff}
````

````python
# Summary table
rows = []
for col in TARGETS:
    r = stationarity_results[col]
    rows.append({
        'Series': col,
        'ADF p (level)':  round(r['level']['adf_p'], 4),
        'KPSS p (level)': round(r['level']['kpss_p'], 4),
        'Stationary at level?': r['level']['adf_stationary'] and r['level']['kpss_stationary'],
        'ADF p (1st diff)':  round(r['diff']['adf_p'], 4),
        'KPSS p (1st diff)': round(r['diff']['kpss_p'], 4),
        'Stationary after diff?': r['diff']['adf_stationary'] and r['diff']['kpss_stationary'],
    })
pd.DataFrame(rows)
````

````python
# ACF / PACF plots
for col in TARGETS:
    series = df_raw[col].dropna()
    print(f"\n{'─'*55}  {col}  (level)")
    plot_acf_pacf(series, name=col + ' (level)')

    diff_series = series.diff().dropna()
    print(f"{'─'*55}  {col}  (1st diff)")
    plot_acf_pacf(diff_series, name=col + ' (1st diff)')
````

````python
# Train-Test Split
splits = {}

for col in TARGETS:
    series = df_raw[col].dropna()
    train = series.iloc[:-TEST_MONTHS]
    test  = series.iloc[-TEST_MONTHS:]
    splits[col] = {'train': train, 'test': test}
    print(f"{col:<22}  train={len(train)}  test={len(test)}  "
          f"({test.index[0].date()} → {test.index[-1].date()})")
````

````python
import warnings
warnings.filterwarnings('ignore')
# ARIMA Grid Search
best_arima_orders = {}

for col in TARGETS:
    print(f"\nGrid search: {col}")
    train = splits[col]['train']
    gs = arima_grid_search(train, p_range=(0, 4), d_range=(0, 2), q_range=(0, 4))
    best_order = gs.iloc[0]['order']
    best_arima_orders[col] = best_order
    print(f"  Best order by AIC: ARIMA{best_order}")
    print(gs.head(5).to_string(index=False))
````

````python
# ARIMA - Fit, Forecast and Evaluate
arima_results  = {}
arima_metrics  = {}

for col in TARGETS:
    train = splits[col]['train']
    test  = splits[col]['test']
    order = best_arima_orders[col]

    print(f"\n{'='*55}")
    print(f" ARIMA{order} — {col}")
    print(f"{'='*55}")

    # Fit
    model = ARIMA(train, order=order)
    fit   = model.fit()
    print(fit.summary())

    # Forecast on test set
    forecast_obj  = fit.get_forecast(steps=TEST_MONTHS)
    forecast_mean = forecast_obj.predicted_mean
    conf_int      = forecast_obj.conf_int()

    # Align index
    forecast_mean.index = test.index
    conf_int.index      = test.index

    # Metrics
    metrics = evaluate_forecast(test, forecast_mean, label=f'ARIMA{order} {col}')
    arima_metrics[col]  = metrics

    # Plot
    plot_forecast(train, test, forecast_mean, conf_int,
                  name=col, model_label=f'ARIMA{order}')

    arima_results[col] = {'fit': fit, 'forecast_mean': forecast_mean, 'conf_int': conf_int}
````

````python
# ARIMA - Residual diagnostics
for col in TARGETS:
    residual_diagnostics(arima_results[col]['fit'],
                         name=f'ARIMA{best_arima_orders[col]} — {col}')
````

````python
# SARIMA - Grid Search for seasonal orders
# Only fit SARIMA for series that show meaningful seasonality (seasonal strength > 0.01 is a loose threshold).
def sarima_grid_search(train, p_range=(0,3), d_range=(0,2), q_range=(0,3),
                       P_range=(0,2), D_range=(0,2), Q_range=(0,2), s=12, verbose=False):
    """Search best SARIMA (p,d,q)(P,D,Q,s) by AIC."""
    results = []
    pdq   = list(itertools.product(range(*p_range), range(*d_range), range(*q_range)))
    PDQs  = list(itertools.product(range(*P_range), range(*D_range), range(*Q_range)))

    total = len(pdq) * len(PDQs)
    print(f"  Evaluating {total} SARIMA combinations...")

    for order in pdq:
        for seasonal_order in PDQs:
            try:
                m = SARIMAX(train,
                            order=order,
                            seasonal_order=(*seasonal_order, s),
                            enforce_stationarity=False,
                            enforce_invertibility=False).fit(disp=False)
                results.append({
                    'order': order,
                    'seasonal_order': (*seasonal_order, s),
                    'AIC': m.aic,
                    'BIC': m.bic
                })
                if verbose:
                    print(f"  SARIMA{order}x{(*seasonal_order, s)}  AIC={m.aic:.2f}")
            except Exception:
                pass

    results_df = pd.DataFrame(results).sort_values('AIC').reset_index(drop=True)
    return results_df
````

````python
best_sarima_orders = {}

for col in TARGETS:
    print(f"\nSARIMA grid search: {col}")
    train = splits[col]['train']
    gs = sarima_grid_search(train,
                            p_range=(0, 3), d_range=(0, 2), q_range=(0, 3),
                            P_range=(0, 2), D_range=(0, 2), Q_range=(0, 2),
                            s=12)
    best_row = gs.iloc[0]
    best_sarima_orders[col] = {
        'order': best_row['order'],
        'seasonal_order': best_row['seasonal_order']
    }
    print(f"  Best: SARIMA{best_row['order']}x{best_row['seasonal_order']}  AIC={best_row['AIC']:.2f}")
    print(gs.head(5).to_string(index=False))
````

````python
# SARIMA - Fit, Forecast and Evaluate
sarima_results = {}
sarima_metrics = {}

for col in TARGETS:
    train = splits[col]['train']
    test  = splits[col]['test']
    order         = best_sarima_orders[col]['order']
    seasonal_order = best_sarima_orders[col]['seasonal_order']

    print(f"\n{'='*55}")
    print(f" SARIMA{order}x{seasonal_order} — {col}")
    print(f"{'='*55}")

    model = SARIMAX(train,
                    order=order,
                    seasonal_order=seasonal_order,
                    enforce_stationarity=False,
                    enforce_invertibility=False)
    fit = model.fit(disp=False)
    print(fit.summary())

    # Forecast
    forecast_obj  = fit.get_forecast(steps=TEST_MONTHS)
    forecast_mean = forecast_obj.predicted_mean
    conf_int      = forecast_obj.conf_int()

    forecast_mean.index = test.index
    conf_int.index      = test.index

    # Metrics
    metrics = evaluate_forecast(test, forecast_mean,
                                label=f'SARIMA{order}x{seasonal_order} {col}')
    sarima_metrics[col] = metrics

    # Plot
    plot_forecast(train, test, forecast_mean, conf_int,
                  name=col, model_label=f'SARIMA{order}x{seasonal_order}')

    sarima_results[col] = {'fit': fit, 'forecast_mean': forecast_mean, 'conf_int': conf_int}
````

````python
#Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import seaborn as sns
from statsmodels.tsa.api import VAR
from statsmodels.tsa.stattools import adfuller,kpss, grangercausalitytests
from statsmodels.tsa.vector_ar.vecm import coint_johansen
from statsmodels.stats.stattools import durbin_watson
from statsmodels.stats.diagnostic import acorr_ljungbox
from sklearn.metrics import mean_absolute_error, mean_squared_error
````

````python
#Data Loading and Variable Engineering
df_var = pd.read_csv("monthly_macro_data_ffill_bfill.csv", parse_dates=['date'])
df_var = df_var.sort_values('date').set_index('date')
df_var.index.freq = 'MS'

#Derive stationary macro variables
df_var['cpi_log_return'] = np.log(df_var['us_cpi_index'] / df_var['us_cpi_index'].shift(1))
df_var['fed_rate_diff'] = df_var['us_fed_funds_rate'].diff()
df_var['yield_10y_diff'] = df_var['us_10y_yield'].diff()
df_var['unemp_diff'] = df_var['us_unemployment_rate'].diff()

#COVID-19 shock dummy (March 2020)
df_var['covid_dummy'] = (df_var.index == '2020-03-01').astype(int)

#VAR variable set (all must be stationary)
VAR_COLS = [
    'usd_inr_monthly_change',
    'brent_monthly_return',
    'fed_rate_diff',
    'yield_10y_diff',
    'cpi_log_return',
    'unemp_diff',
    'sp500_monthly_return',
    'nifty50_monthly_return',
    'gold_monthly_return',
]

var_df = df_var[VAR_COLS].dropna()
print(f"VAR dataset: {var_df.shape[0]} observations × {var_df.shape[1]} variables")
print(f"Period     : {var_df.index[0].date()} → {var_df.index[-1].date()}")
var_df.head()
````

````python
#Stationarity
def stationarity_report(df, alpha=0.05):
    rows = []
    for col in df.columns:
        s = df[col].dropna()
        adf_stat, adf_p, _, _, _, _ = adfuller(s, autolag='AIC')
        kpss_stat, kpss_p, _, _     = kpss(s, regression='c', nlags='auto')
        adf_ok  = adf_p  < alpha
        kpss_ok = kpss_p > alpha
        verdict = 'Stationary' if (adf_ok and kpss_ok) else 'Conflicting' if (adf_ok or kpss_ok) else 'Non-Stationary'
        rows.append({'Variable': col,
                     'ADF Stat': round(adf_stat, 4), 'ADF p': round(adf_p, 4),
                     'KPSS Stat': round(kpss_stat, 4), 'KPSS p': round(kpss_p, 4),
                     'Verdict': verdict})
    return pd.DataFrame(rows).set_index('Variable')

stat_table = stationarity_report(var_df)
print("\nStationarity Summary (ADF + KPSS):\n")
print(stat_table.to_string())
````

````python
#Johansen Cointegration Test(close prices + macro levels)
#Tests whether long-run equilibrium relationships exist among the I(1) level series. If cointegration is found among levels, a VECM would be warranted. If not, VAR on stationary (return/differenced) series is appropriate.
level_cols = [
    'usd_inr_close', 'brent_close', 'us_fed_funds_rate',
    'us_10y_yield', 'us_cpi_index', 'sp500_close', 'nifty50_close', 'gold_close'
]
df_levels = df_var[level_cols].dropna()

joh = coint_johansen(df_levels, det_order=0, k_ar_diff=1)

print("=" * 65)
print("JOHANSEN COINTEGRATION TEST  (det_order=0, k_ar_diff=1)")
print("=" * 65)

trace_stat = joh.lr1
trace_cv   = joh.cvt          # columns: 90%, 95%, 99%
max_stat   = joh.lr2
max_cv     = joh.cvm

print("\nTrace Statistic:")
print(f"  {'H0: rank ≤ r':<15}  {'Trace':>10}  {'CV 5%':>10}  {'Result'}")
print("  " + "-" * 55)
for r in range(len(trace_stat)):
    sig = 'Reject H0' if trace_stat[r] > trace_cv[r, 1] else '  Fail to Reject'
    print(f"  r ≤ {r:<10}  {trace_stat[r]:>10.4f}  {trace_cv[r,1]:>10.4f}  {sig}")

print("\nMax-Eigenvalue Statistic:")
print(f"  {'H0: rank ≤ r':<15}  {'MaxEig':>10}  {'CV 5%':>10}  {'Result'}")
print("  " + "-" * 55)
for r in range(len(max_stat)):
    sig = 'Reject H0' if max_stat[r] > max_cv[r, 1] else '  Fail to Reject'
    print(f"  r ≤ {r:<10}  {max_stat[r]:>10.4f}  {max_cv[r,1]:>10.4f}  {sig}")

print("\n→ Cointegration found → long-run relationships exist among level series.")
print("  VAR fitted on stationary (return/differenced) variables as per methodology.")
````

````python
#Granger Causality Test
MAX_LAGS = 6
ALPHA = 0.05
TARGET = 'nifty50_monthly_return'

variables = var_df.columns.tolist()
granger_records = []

for cause in variables:
    for effect in variables:
        if cause == effect:
            continue
        pair = var_df[[effect, cause]].dropna()
        res  = grangercausalitytests(pair, maxlag=MAX_LAGS, verbose=False)
        p_vals = [res[lag][0]['ssr_ftest'][1] for lag in range(1, MAX_LAGS + 1)]
        min_p  = min(p_vals)
        best_lag = p_vals.index(min_p) + 1
        granger_records.append({
            'Cause': cause, 'Effect': effect,
            'Min p-value': round(min_p, 4), 'Best Lag': best_lag,
            'Granger Causes?': 'Yes' if min_p < ALPHA else 'No'
        })

granger_full_df = pd.DataFrame(granger_records)

# Highlight results involving the primary target
target_granger = granger_full_df[granger_full_df['Effect'] == TARGET].copy()
target_granger = target_granger.sort_values('Min p-value')

print(f"\nGranger Causality → Effect on {TARGET}")
print("=" * 75)
print(target_granger[['Cause', 'Min p-value', 'Best Lag', 'Granger Causes?']].to_string(index=False))
````

````python
#Heatmap
pivot = granger_full_df.pivot(index='Cause', columns='Effect', values='Min p-value')

fig, axes = plt.subplots(1, 2, figsize=(18, 6))

# p-value heatmap
sns.heatmap(pivot, annot=True, fmt='.3f', cmap='RdYlGn_r',
            vmin=0, vmax=0.1, linewidths=0.4, ax=axes[0],
            cbar_kws={'label': 'Min p-value (F-test)'})
axes[0].set_title(f'Granger Causality — p-value Matrix\n(green = significant, red = not)',
                  fontsize=12, fontweight='bold')
axes[0].set_xlabel('Effect Variable'); axes[0].set_ylabel('Cause Variable')
axes[0].tick_params(axis='x', rotation=30)
axes[0].tick_params(axis='y', rotation=0)

# Binary significance heatmap
sig_pivot = (pivot < ALPHA).astype(int)
sns.heatmap(sig_pivot, annot=True, fmt='d', cmap='RdYlGn',
            vmin=0, vmax=1, linewidths=0.4, ax=axes[1],
            cbar_kws={'label': f'1 = p < {ALPHA} (Granger Causes)'})
axes[1].set_title(f'Granger Causality — Significance Matrix\n(1 = Granger Causes at α={ALPHA})',
                  fontsize=12, fontweight='bold')
axes[1].set_xlabel('Effect Variable'); axes[1].set_ylabel('Cause Variable')
axes[1].tick_params(axis='x', rotation=30)
axes[1].tick_params(axis='y', rotation=0)

plt.suptitle('Pairwise Granger Causality Test Results', fontsize=14, fontweight='bold', y=1.01)
plt.tight_layout()
plt.show()
````

````python
#VAR Lag Order Selection
TEST_MONTHS = 12

train_var = var_df.iloc[:-TEST_MONTHS]
test_var  = var_df.iloc[-TEST_MONTHS:]

print(f"Train: {train_var.index[0].date()} → {train_var.index[-1].date()}  ({len(train_var)} obs)")
print(f"Test : {test_var.index[0].date()}  → {test_var.index[-1].date()}  ({len(test_var)} obs)")

var_selector = VAR(train_var)
lag_sel      = var_selector.select_order(maxlags=MAX_LAGS)

print("\n")
print(lag_sel.summary())

aic_lag  = int(lag_sel.selected_orders['aic'])
bic_lag  = int(lag_sel.selected_orders['bic'])
hqic_lag = int(lag_sel.selected_orders['hqic'])

# Override if lag = 0
optimal_lag = aic_lag if aic_lag > 0 else (bic_lag if bic_lag > 0 else 1)

print(f"\nSelected Lag Orders — AIC: {aic_lag}  |  BIC: {bic_lag}  |  HQIC: {hqic_lag}")
print(f"→ Optimal lag used for modelling: p = {optimal_lag}  (AIC criterion)")
````

Optimal lag length is selected using three information criteria. AIC is used as the primary criterion; BIC is noted for comparison as it penalises complexity more heavily.

````python
#Fit VAR(p) Model
var_fitted = var_selector.fit(optimal_lag)
print(var_fitted.summary())
````

````python
#Residual Diagnostics
#White noise residuals confirm model adequacy.
from statsmodels.graphics.gofplots import qqplot
from statsmodels.stats.diagnostic import het_arch # Import for ARCH test

resid = var_fitted.resid

#Durbin-Watson
dw_vals = durbin_watson(resid)
print("Durbin-Watson Statistics (should be ≈ 2.0 for no autocorrelation):")
print("-" * 55)
for col, dw in zip(var_df.columns, dw_vals):
    flag = '--' if 1.5 < dw < 2.5 else 'x'
    print(f" {flag}  {col:<35}  DW = {dw:.4f}")

#Ljung-Box on target equation
target_resid = resid[TARGET]
lb = acorr_ljungbox(target_resid, lags=[6, 12, 18], return_df=True)
print(f"\nLjung-Box Test — {TARGET}:")
print(lb.to_string())
wn = (lb['lb_pvalue'] > ALPHA).all()
print("\n→", "Residuals are white noise (no autocorrelation)" if wn
      else "Residuals show autocorrelation — consider higher lag")

#ARCH test on target equation
arch_stat, arch_p, _, _ = het_arch(target_resid, nlags=6)
print(f"\nARCH Test — {TARGET}  (LM stat={arch_stat:.4f}, p={arch_p:.4f})")
print("→", "No ARCH effects" if arch_p > ALPHA else "ARCH effects present")
````

````python
#Impulse Response Functions
#IRFs show how a one standard deviation orthogonal shock to each predictor propagates into usd_inr_monthly_change over 12 months. Cholesky decomposition is used for orthogonalisation. This directly tests (Fed Rate shock → INR depreciation within 1–3 months).
irf = var_fitted.irf(periods=24)

#Target variable (e.g., USD/INR)
target_idx = list(var_df.columns).index(TARGET)

n_vars = len(var_df.columns)
n_cols = 3
n_rows = (n_vars + n_cols - 1) // n_cols

fig, axes = plt.subplots(n_rows, n_cols, figsize=(18, n_rows * 4))
axes = axes.flatten()

for i, impulse_var in enumerate(var_df.columns):

    impulse_idx = list(var_df.columns).index(impulse_var)

    # IRF values
    irf_vals = irf.orth_irfs[:, target_idx, impulse_idx]

    # Confidence intervals (approx using stderr)
    try:
        stderr = irf.stderr(orth=True)
        lower = irf_vals - 1.96 * stderr[:, target_idx, impulse_idx]
        upper = irf_vals + 1.96 * stderr[:, target_idx, impulse_idx]

        axes[i].fill_between(range(len(irf_vals)), lower, upper, alpha=0.2)
    except:
        pass

    #Plot IRF
    axes[i].plot(range(len(irf_vals)), irf_vals, linewidth=2)

    #Reference lines
    axes[i].axhline(0, linestyle='--', linewidth=0.8)
    axes[i].axvline(3, linestyle=':', linewidth=0.8)

    axes[i].set_title(f'Shock: {impulse_var}', fontsize=10)
    axes[i].set_xlabel('Months')
    axes[i].set_ylabel('Response')

#Hide extra subplots
for j in range(i + 1, len(axes)):
    axes[j].set_visible(False)

plt.suptitle(f'IRF: Response of {TARGET} to Shocks', fontsize=14)
plt.tight_layout()
plt.show()
````

````python
#Out of Sample Forecasting & Evaluation
#Recursive rolling forecast over test period
#At each step, forecast 1-step ahead using all available history (more realistic)
rolling_preds = {col: [] for col in var_df.columns}

full_data = var_df.copy()

for t in range(TEST_MONTHS):
    history = full_data.iloc[:len(train_var) + t]
    m       = VAR(history).fit(optimal_lag)
    fc_1    = m.forecast(history.values[-optimal_lag:], steps=1)[0]
    for j, col in enumerate(var_df.columns):
        rolling_preds[col].append(fc_1[j])

rolling_fc_df = pd.DataFrame(rolling_preds, index=test_var.index)

print(f"Rolling 1-step-ahead forecast generated: {len(rolling_fc_df)} steps\n")

#Evaluation metrics
def directional_accuracy(actual, predicted):
    a = np.array(actual); p = np.array(predicted)
    return np.mean(np.sign(a[1:]) == np.sign(p[1:])) * 100

eval_rows = []
print(f"{'Variable':<32}  {'RMSE':>8}  {'MAE':>8}  {'MAPE%':>8}  {'Dir Acc%':>9}")
print("-" * 75)
for col in var_df.columns:
    actual    = test_var[col].values
    predicted = rolling_fc_df[col].values
    rmse = np.sqrt(mean_squared_error(actual, predicted))
    mae  = mean_absolute_error(actual, predicted)
    #MAPE: guard against zero denominators
    nz   = np.abs(actual) > 1e-8
    mape = np.mean(np.abs((actual[nz] - predicted[nz]) / actual[nz])) * 100 if nz.sum() > 0 else np.nan
    da   = directional_accuracy(actual, predicted)
    eval_rows.append({'Variable': col, 'RMSE': rmse, 'MAE': mae, 'MAPE (%)': mape,
                      'Dir Accuracy (%)': da})
    flag = '-' if col == TARGET else ' '
    print(f"  {flag} {col:<30}  {rmse:>8.4f}  {mae:>8.4f}  {mape:>8.2f}  {da:>9.1f}%")

eval_df = pd.DataFrame(eval_rows).set_index('Variable')
````

# MODULE 4: Macro-Driven ML Return Prediction

## 🎯 What Is This Module Trying to Do?

The central question: **Can lagged macroeconomic variables predict whether stock markets will go UP or DOWN next month?**

Instead of predicting the *exact return value* (like Module 2's ARIMA did), Module 4 simplifies the problem into a **binary classification** task:

| Label | Meaning |
|-------|--------|
| **1 (Up)** | Next month's return is **positive** |
| **0 (Down)** | Next month's return is **negative or zero** |

This is done for two equity indices: **Nifty50** (India) and **S&P500** (USA).


##### This module builds a binary classification framework to predict whether Nifty50 and S&P500 monthly returns will be positive (Up = 1) or negative (Down = 0), using lagged macroeconomic features and strict time-series cross-validation to prevent data leakage.

````python
# Loading essential libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.model_selection import TimeSeriesSplit
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (accuracy_score, precision_score, recall_score,
                             f1_score, roc_auc_score, confusion_matrix,
                             ConfusionMatrixDisplay, roc_curve)
````

````python
# Loading clean dataset
df_ml = pd.read_csv('monthly_macro_data_ffill_bfill.csv', parse_dates=['date'])
df_ml = df_ml.sort_values('date').reset_index(drop=True)
df_ml = df_ml.set_index('date')
df_ml.index.freq = 'MS'

print(f"Dataset shape  : {df_ml.shape}")
print(f"Date range     : {df_ml.index.min().date()} → {df_ml.index.max().date()}")
print(f"Columns        : {df_ml.columns.tolist()}")
````

````python
# Feature engineering
fe = df_ml.copy()

# A: CPI month-on-month change (log difference)
fe['cpi_mom'] = np.log(fe['us_cpi_index'] / fe['us_cpi_index'].shift(1))

# B: Lag 1, 2, 3 for each macro driver
lag_ft = [
    'us_fed_funds_rate',
    'cpi_mom',
    'us_10y_yield',
    'us_unemployment_rate',
    'brent_monthly_return',
    'usd_inr_monthly_change',
]

for col in lag_ft:
    for lag in [1, 2, 3]:
        fe[f'{col}_lag{lag}'] = fe[col].shift(lag)

# C: Yield spread — 10Y yield minus Fed funds rate (curve steepness)
fe['yield_spread'] = fe['us_10y_yield'] - fe['us_fed_funds_rate']
fe['yield_spread_lag1'] = fe['yield_spread'].shift(1)
fe['yield_spread_lag2'] = fe['yield_spread'].shift(2)

# D: Gold momentum — 3-month rolling return on gold
fe['gold_mom_3m'] = fe['gold_monthly_return'].rolling(3).sum()
fe['gold_mom_3m_lag1'] = fe['gold_mom_3m'].shift(1)

# E: Equity momentum — lagged returns of the target itself
fe['nifty50_ret_lag1']  = fe['nifty50_monthly_return'].shift(1)
fe['nifty50_ret_lag2']  = fe['nifty50_monthly_return'].shift(2)
fe['sp500_ret_lag1']    = fe['sp500_monthly_return'].shift(1)
fe['sp500_ret_lag2']    = fe['sp500_monthly_return'].shift(2)

# Final feature list
ft_cols = (
    [f'{col}_lag{lag}' for col in lag_ft for lag in [1, 2, 3]] +
    ['yield_spread_lag1', 'yield_spread_lag2',
     'gold_mom_3m_lag1',
     'nifty50_ret_lag1', 'nifty50_ret_lag2',
     'sp500_ret_lag1',   'sp500_ret_lag2']
)

print(f"Total features constructed : {len(ft_cols)}")
print(f"Feature list:\n  " + "\n  ".join(ft_cols))
````

## 🔧 Feature Engineering — The 25 Input Variables

**The golden rule:** All 25 features are **strictly lagged** — no future information is ever used. This prevents **look-ahead bias** (the model cannot "cheat" by seeing the same month's macro data it is trying to predict).

```
Total features constructed : 25
```

| Brent Monthly Return | ✓ | ✓ | ✓ |
| USD/INR Monthly Change | ✓ | ✓ | ✓ |

> **Why multiple lags?** Financial markets don't react instantly. A Fed rate hike today may take 2–3 months to slow consumer spending and affect equity sentiment. Lag 1, 2, 3 captures these delayed effects.

### Feature Group B — Yield Spread (2 Features)
```python
yield_spread = us_10y_yield - us_fed_funds_rate
```
Captured at lag 1 and lag 2.  
> **Economic meaning:** A *wide positive* spread = market expects future growth (bullish for equities). An *inverted/narrow* spread = recession signal (bearish). This is one of the most reliable macro indicators in finance.

```python
gold_mom_3m = gold_monthly_return.rolling(3).sum()  # then shifted by 1
```

- `nifty50_ret_lag1`, `nifty50_ret_lag2`
- `sp500_ret_lag1`, `sp500_ret_lag2`

> The idea: markets exhibit **momentum** — recent positive returns may continue short-term. Including the target's own lagged return tests this hypothesis.

When predicting **"what will happen in May"**, you can only legally use:
- April data → Lag 1
- March data → Lag 2  
- February data → Lag 3

The notebook strictly enforces this with `.shift(lag)` — ensuring zero future data leakage.

---
## 🧪 Validation Strategy — Walk-Forward Time-Series Cross-Validation

### Why This Matters

**The training set always stays in the past; the test set is always the future.** The model is never evaluated on data it could have seen during training. This replicates real-world conditions.

| ❌ Standard K-Fold (WRONG for time series) | ✅ TimeSeriesSplit (CORRECT) |
|------------------------------------------|-----------------------------|
| Randomly shuffles data | Preserves chronological order |
| Test data can be earlier than training | Test data is always after training |
| Inflated, unrealistic performance | Conservative, honest performance |
| Data leakage from future to past | Zero data leakage |

> **No shuffling is done** — which is the correct and required approach for time series. Each fold's training window expands (walk-forward), and each test window covers a distinct ~2.5-year period.

### Objective achieved: Strict walk-forward cross-validation

````python
# function to run CV
def run_cv(model, X, y, tscv, model_name, target_name, scale=False):
    fold_metrics = []
    all_true, all_pred, all_prob = [], [], []

    for fold, (train_idx, test_idx) in enumerate(tscv.split(X), 1):
        X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
        y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]

        if scale:
            scaler  = StandardScaler()
            X_train = scaler.fit_transform(X_train)
            X_test  = scaler.transform(X_test)

        model.fit(X_train, y_train)
        y_pred = model.predict(X_test)
        y_prob = (model.predict_proba(X_test)[:, 1]
                  if hasattr(model, 'predict_proba') else y_pred.astype(float))

        acc  = accuracy_score(y_test, y_pred)
        prec = precision_score(y_test, y_pred, zero_division=0)
        rec  = recall_score(y_test, y_pred, zero_division=0)
        f1   = f1_score(y_test, y_pred, zero_division=0)
        try:
            auc = roc_auc_score(y_test, y_prob)
        except ValueError:
            auc = np.nan

        fold_metrics.append({
            'Fold': fold, 'Train size': len(train_idx), 'Test size': len(test_idx),
            'Accuracy': round(acc, 4), 'Precision': round(prec, 4),
            'Recall': round(rec, 4), 'F1': round(f1, 4), 'ROC-AUC': round(auc, 4),
        })
        all_true.extend(y_test.tolist())
        all_pred.extend(y_pred.tolist())
        all_prob.extend(y_prob.tolist())

        print(f"  Fold {fold}: Acc={acc:.3f}  Prec={prec:.3f}  Rec={rec:.3f}  "
              f"F1={f1:.3f}  AUC={auc:.3f}")

    results_df = pd.DataFrame(fold_metrics)
    numeric_cols = ['Accuracy', 'Precision', 'Recall', 'F1', 'ROC-AUC']
    means = results_df[numeric_cols].mean().round(4)
    stds  = results_df[numeric_cols].std().round(4)
    print(f"\n  Mean: " + "  ".join(f"{c}={means[c]:.3f}±{stds[c]:.3f}" for c in numeric_cols))

    all_preds = {
        'y_true': np.array(all_true),
        'y_pred': np.array(all_pred),
        'y_prob': np.array(all_prob),
    }
    return results_df, all_preds
````

## 📐 Metric Definitions

| Metric | Formula | Plain Language |
|--------|---------|----------------|
| **Accuracy** | (TP + TN) / Total | Out of ALL predictions, how many were correct? |
| **Precision** | TP / (TP + FP) | When the model predicted "Up", how often was it actually Up? |
| **Recall** | TP / (TP + FN) | Out of all actual "Up" months, how many did the model catch? |
| **F1 Score** | 2 × (P × R) / (P + R) | Harmonic mean of Precision and Recall — use this when classes are imbalanced |
| **ROC-AUC** | Area under ROC curve | Can the model rank Up vs Down months correctly by probability? 0.5=random, 1.0=perfect |

The notebook produces a 2×2 grid of plots (Top-10 RF MDI Importance vs Top-10 LR |Coefficient| for each target). This cross-validates which features **both models agree** are most informative.

**When RF and LR agree on a feature's importance**, that feature has a robust claim to being a genuine predictor — one model's non-linear tree splits and another model's linear separation both depend on it.

````python
# ─────────────────────────────────────────────────────────────────────────────
# SECTION 0: IMPORTS
# All libraries needed for the entire module are loaded here up front.
# ─────────────────────────────────────────────────────────────────────────────
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import matplotlib.patches as mpatches
import seaborn as sns
import warnings
from scipy import stats              # for Fisher Z-transformation and t-tests
from itertools import combinations  # to generate all asset-pair combos

warnings.filterwarnings('ignore')

# Plot aesthetics — consistent with previous modules
plt.rcParams['figure.figsize'] = (14, 5)
plt.rcParams['axes.grid']      = True
plt.rcParams['grid.alpha']     = 0.3
plt.rcParams['font.size']      = 10
````

````python
# ─────────────────────────────────────────────────────────────────────────────
# SECTION 1: DATA LOADING AND PREPARATION
# We use the ffill/bfill cleaned CSV produced by Module 1 (Person 1).
# This ensures consistency with all upstream modules.
# ─────────────────────────────────────────────────────────────────────────────

print("=" * 65)
print(" MODULE 5: CROSS-MARKET CORRELATION ANALYSIS")
print("=" * 65)

# Load the clean, imputed dataset created by Person 1
# The ffill+bfill strategy filled the 1 missing row in us_cpi_index, etc.
df = pd.read_csv('monthly_macro_data_ffill_bfill.csv', parse_dates=['date'])
df = df.sort_values('date').reset_index(drop=True)
df = df.set_index('date')
df.index.freq = 'MS'   # Month-Start frequency (needed for rolling calcs)

print(f"\nDataset loaded : {df.shape[0]} observations × {df.shape[1]} columns")
print(f"Date range     : {df.index.min().date()} → {df.index.max().date()}")

# ── Return columns we will use throughout this module ────────────────────────
# We work on monthly returns (log returns / pct changes), NOT raw closes,
# because returns are stationary and correlations on levels are spurious
# (high just because both trends upward over time).
RETURN_COLS = [
    'nifty50_monthly_return',   # Indian equity log return
    'sp500_monthly_return',     # US equity log return
    'gold_monthly_return',      # Gold log return
    'brent_monthly_return',     # Brent crude log return
    'usd_inr_monthly_change',   # USD/INR pct change (positive = INR weakens)
]

# Friendly short labels for plot axes (less cluttered)
LABELS = {
    'nifty50_monthly_return': 'Nifty50',
    'sp500_monthly_return':   'S&P500',
    'gold_monthly_return':    'Gold',
    'brent_monthly_return':   'Brent',
    'usd_inr_monthly_change': 'USD/INR',
}

# Keep only the return columns and drop the one NaN row (first observation)
ret_df = df[RETURN_COLS].dropna()

print(f"\nReturn series  : {ret_df.shape[0]} complete observations")
print(f"Period         : {ret_df.index[0].date()} → {ret_df.index[-1].date()}")
print(f"\nReturn column stats:")
print(ret_df.describe().round(4))

````

````python
# ─────────────────────────────────────────────────────────────────────────────
# SECTION 2: FULL-PERIOD STATIC CORRELATION MATRIX
# This is the baseline — the single unconditional correlation across the
# entire 2010–2025 window.  It tells us the "average" relationship before
# we break it into regimes or roll it over time.
# ─────────────────────────────────────────────────────────────────────────────

print("\n" + "=" * 65)
print(" 2. FULL-PERIOD STATIC CORRELATION MATRIX (2010–2025)")
print("=" * 65)

# Rename columns to short labels for readability in plots
ret_plot = ret_df.rename(columns=LABELS)

# Pearson correlation (linear, pairwise)
full_corr = ret_plot.corr()
print("\nFull-period Pearson correlation matrix:")
print(full_corr.round(4))

# ── Heatmap: full-period correlation ─────────────────────────────────────────
fig, ax = plt.subplots(figsize=(8, 6))
mask = np.triu(np.ones_like(full_corr, dtype=bool), k=1)  # hide upper triangle
sns.heatmap(
    full_corr,
    annot=True,          # show correlation numbers inside cells
    fmt='.2f',
    cmap='RdYlGn',       # red = strong negative, green = strong positive
    vmin=-1, vmax=1,
    linewidths=0.5,
    ax=ax,
    mask=mask,           # show lower triangle only (symmetric matrix)
    annot_kws={'size': 11, 'weight': 'bold'},
)
ax.set_title(
    'Full-Period Correlation Matrix — Monthly Returns (Jan 2010 – Dec 2025)',
    fontsize=12, fontweight='bold', pad=14
)
plt.tight_layout()
plt.show()

# ── Interpretation printout ───────────────────────────────────────────────────
print("\nKey full-period correlations:")
pairs = list(combinations(ret_plot.columns, 2))
for a, b in pairs:
    r = full_corr.loc[a, b]
    print(f"  {a:<10} vs {b:<10}  r = {r:+.4f}")

````

````python
# ─────────────────────────────────────────────────────────────────────────────
# SECTION 3: ROLLING CORRELATIONS (12-MONTH AND 24-MONTH WINDOWS)
#
# A single static correlation collapses the entire 15-year period into one
# number.  Rolling windows reveal HOW the correlation evolves — particularly
# useful for spotting regime changes during COVID (2020), the rate-hike
# cycle (2022–23), etc.
#
# Objective 6a: "Rolling 12-month and 24-month Pearson correlation between
#                all asset pairs"
# ─────────────────────────────────────────────────────────────────────────────

print("\n" + "=" * 65)
print(" 3. ROLLING CORRELATIONS (12M AND 24M WINDOWS)")
print("=" * 65)

WINDOWS = {'12M Rolling': 12, '24M Rolling': 24}

# ── 3a. Compute rolling correlations for ALL pairs ────────────────────────────
# We store them in a dict keyed by (window_label, col_a, col_b)
rolling_corr = {}
for wlabel, wsize in WINDOWS.items():
    for a, b in pairs:
        key = (wlabel, a, b)
        rolling_corr[key] = (
            ret_plot[a]
            .rolling(window=wsize, min_periods=wsize)
            .corr(ret_plot[b])
        )

# ── 3b. Plot all rolling correlations for 12M and 24M side-by-side ───────────
# Each subplot = one asset pair, two lines (12M vs 24M)
n_pairs  = len(pairs)
n_cols   = 2
n_rows   = (n_pairs + 1) // n_cols

fig, axes = plt.subplots(n_rows, n_cols, figsize=(18, n_rows * 3.5))
axes      = axes.flatten()

for idx, (a, b) in enumerate(pairs):
    ax = axes[idx]
    for wlabel, wsize in WINDOWS.items():
        key   = (wlabel, a, b)
        color = 'steelblue' if wsize == 12 else 'darkorange'
        ax.plot(rolling_corr[key].index, rolling_corr[key].values,
                label=wlabel, color=color, linewidth=1.5, alpha=0.85)
    ax.axhline(0, color='gray', linewidth=0.8, linestyle=':')
    ax.axhline(full_corr.loc[a, b], color='crimson', linewidth=0.9,
               linestyle='--', alpha=0.6, label='Full-period avg')
    ax.set_title(f'{a} vs {b}', fontsize=10, fontweight='bold')
    ax.set_ylim(-1.05, 1.05)
    ax.legend(fontsize=7)
    ax.xaxis.set_major_formatter(mdates.DateFormatter('%Y'))
    ax.tick_params(axis='x', rotation=30, labelsize=8)

# Hide any empty subplots if pairs count is odd
for idx in range(n_pairs, len(axes)):
    axes[idx].set_visible(False)

fig.suptitle('Rolling Pearson Correlations — All Asset Pairs (12M and 24M Windows)',
             fontsize=13, fontweight='bold')
plt.tight_layout()
plt.show()

````

````python
# ─────────────────────────────────────────────────────────────────────────────
````
````python
# SECTION 5: REGIME-SPECIFIC CORRELATION HEATMAPS
#
# We slice the data into four economic regimes and compute a separate
# correlation matrix for each.  This reveals structural breaks — relationships
# that hold in normal times may invert during crises.
#
# Objective 6b: "Correlation heatmap snapshots: pre-2020, 2020–2021 (COVID),
#                2022–2023 (rate hikes), 2024–2025"
# ─────────────────────────────────────────────────────────────────────────────

print("\n" + "=" * 65)
print(" 5. REGIME-SPECIFIC CORRELATION HEATMAPS")
print("=" * 65)

# Define four regimes
REGIMES = {
    'Pre-COVID\n(2010–2019)': ('2010-01-01', '2019-12-31'),
    'COVID Era\n(2020–2021)': ('2020-01-01', '2021-12-31'),
    'Rate-Hike Cycle\n(2022–2023)': ('2022-01-01', '2023-12-31'),
    'Post-Hike\n(2024–2025)': ('2024-01-01', '2025-12-31'),
}

regime_corr = {}   # store each regime's correlation matrix
regime_n    = {}   # number of observations per regime

for regime_label, (start, end) in REGIMES.items():
    # Slice the return dataframe for this regime window
    subset = ret_plot.loc[start:end].dropna()
    regime_corr[regime_label] = subset.corr()
    regime_n[regime_label]    = len(subset)
    print(f"  {regime_label.replace(chr(10), ' '):<30}  "
          f"n = {len(subset)} months")

# ── Plot four heatmaps side-by-side ──────────────────────────────────────────
fig, axes = plt.subplots(1, 4, figsize=(26, 6))

for ax, (regime_label, corr_mat) in zip(axes, regime_corr.items()):
    mask = np.triu(np.ones_like(corr_mat, dtype=bool), k=1)
    sns.heatmap(
        corr_mat, annot=True, fmt='.2f',
        cmap='RdYlGn', vmin=-1, vmax=1,
        linewidths=0.5, ax=ax, mask=mask,
        annot_kws={'size': 9},
        cbar=False,
    )
    n_obs = regime_n[regime_label]
    ax.set_title(f'{regime_label}\n(n={n_obs})',
                 fontsize=10, fontweight='bold')
    ax.tick_params(axis='x', rotation=45, labelsize=8)
    ax.tick_params(axis='y', rotation=0, labelsize=8)

fig.suptitle('Correlation Heatmaps by Economic Regime', fontsize=14,
             fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()

# ── Print correlation change table ───────────────────────────────────────────
print("\nCorrelation change across regimes — key pairs:")
print(f"{'Pair':<22}", end='')
for rl in REGIMES.keys():
    print(f"  {rl.split(chr(10))[0]:>15}", end='')
print()
print("-" * 88)

for a, b in pairs:
    print(f"  {a} vs {b:<12}", end='')
    for rl in REGIMES.keys():
        r = regime_corr[rl].loc[a, b]
        print(f"  {r:>+15.3f}", end='')
    print()
````

````python
# ─────────────────────────────────────────────────────────────────────────────
# SECTION 6: FISHER Z-TRANSFORMATION TEST
#
# Eyeballing different correlation numbers is not enough — we need to test
# whether the differences across regimes are statistically significant.
#
# Fisher Z-transformation converts a Pearson r into a normally distributed
# Z-score.  We then compute a z-test on the difference between two Z-scores
# from two different samples (regimes).
#
# Formula:
#   Z(r) = 0.5 * ln((1+r)/(1-r))
#   SE   = 1 / sqrt(n-3)
#   z_stat = (Z1 - Z2) / sqrt(1/(n1-3) + 1/(n2-3))
#   p_val  = 2 * (1 - Φ(|z_stat|))  [two-tailed]
#
# Objective 6e: "Statistical test: are correlations significantly different
#                across regimes? (use Fisher Z-transformation)"
# ─────────────────────────────────────────────────────────────────────────────

print("\n" + "=" * 65)
print(" 6. FISHER Z-TRANSFORMATION: REGIME SIGNIFICANCE TESTS")
print("=" * 65)

def fisher_z_test(r1, n1, r2, n2):
    """
    Two-sample Fisher Z-test for equality of two Pearson correlations.
    Returns: z_statistic, p_value (two-tailed)
    Note: clips r to ±0.9999 to avoid division-by-zero at r = ±1.
    """
    r1 = np.clip(r1, -0.9999, 0.9999)
    r2 = np.clip(r2, -0.9999, 0.9999)
    z1 = 0.5 * np.log((1 + r1) / (1 - r1))   # Fisher Z of r1
    z2 = 0.5 * np.log((1 + r2) / (1 - r2))   # Fisher Z of r2
    se = np.sqrt(1 / (n1 - 3) + 1 / (n2 - 3))  # pooled standard error
    z_stat = (z1 - z2) / se
    p_val  = 2 * (1 - stats.norm.cdf(abs(z_stat)))  # two-tailed p-value
    return z_stat, p_val

ALPHA = 0.05  # significance threshold

# We test each asset pair for three regime-transitions:
# Pre-COVID → COVID, COVID → Rate Hikes, Rate Hikes → Post-Hike
regime_list  = list(REGIMES.keys())
regime_pairs = [
    (regime_list[0], regime_list[1]),   # Pre-COVID vs COVID
    (regime_list[1], regime_list[2]),   # COVID vs Rate Hikes
    (regime_list[2], regime_list[3]),   # Rate Hikes vs Post-Hike
    (regime_list[0], regime_list[2]),   # Pre-COVID vs Rate Hikes (big jump)
]

fisher_records = []

for a, b in pairs:
    for r1_label, r2_label in regime_pairs:
        r1     = regime_corr[r1_label].loc[a, b]
        r2     = regime_corr[r2_label].loc[a, b]
        n1     = regime_n[r1_label]
        n2     = regime_n[r2_label]
        z_stat, p_val = fisher_z_test(r1, n1, r2, n2)
        fisher_records.append({
            'Asset Pair':       f'{a} vs {b}',
            'Regime A':         r1_label.replace('\n', ' '),
            'Regime B':         r2_label.replace('\n', ' '),
            'r(A)':             round(r1, 4),
            'r(B)':             round(r2, 4),
            'Δr':               round(r2 - r1, 4),
            'Z-stat':           round(z_stat, 4),
            'p-value':          round(p_val, 4),
            'Significant?':     'YES ***' if p_val < ALPHA else 'No',
        })

fisher_df = pd.DataFrame(fisher_records)

print("\nFisher Z-test results (α = 0.05):")
print("=" * 100)
print(fisher_df.to_string(index=False))

# ── Focus: significant regime shifts ─────────────────────────────────────────
sig = fisher_df[fisher_df['Significant?'] == 'YES ***']
print(f"\nStatistically significant regime changes: {len(sig)} out of {len(fisher_df)}")
if len(sig) > 0:
    print(sig[['Asset Pair', 'Regime A', 'Regime B', 'r(A)', 'r(B)', 'Δr',
               'p-value']].to_string(index=False))
else:
    print("  No pair-regime comparison reached p < 0.05 — "
          "likely due to small per-regime sample sizes (n ≈ 24).")
    print("  Directional Δr values still carry economic interpretation.")
````

````python
# ─────────────────────────────────────────────────────────────────────────────
# SECTION 7: SAFE-HAVEN DYNAMICS — USD/INR vs GOLD
#
# A core question in cross-market analysis for Indian markets:
#   "Does the INR weaken when gold rises, and vice versa?"
#
# Economic logic:
#   • In a risk-off episode, global investors flee to gold (safe-haven).
#   • At the same time, capital flows OUT of emerging markets → INR falls.
#   • So gold ↑ should coincide with USD/INR ↑ (INR weakens against USD).
#   • This means Gold returns and USD/INR changes should be POSITIVELY
#     correlated — both rise in risk-off mode.
#
# However, gold is also priced in USD, so when USD strengthens, gold (in USD)
# can fall, creating a NEGATIVE correlation.  The actual direction is
# empirical — we check both the scatter and rolling evolution.
#
# Objective 6d: "USD/INR correlation with gold: safe-haven dynamics —
#               does INR weaken when gold rises?"
# ─────────────────────────────────────────────────────────────────────────────

print("\n" + "=" * 65)
print(" 7. SAFE-HAVEN DYNAMICS: USD/INR vs GOLD")
print("=" * 65)

gold_col    = 'gold_monthly_return'
usd_inr_col = 'usd_inr_monthly_change'

# Rename for convenience
gold_ret   = ret_df[gold_col].rename('Gold Return')
usdinr_ret = ret_df[usd_inr_col].rename('USD/INR Change')

# Full-period Pearson r and p-value
r_full, p_full = stats.pearsonr(
    gold_ret.dropna(), usdinr_ret.loc[gold_ret.dropna().index]
)
print(f"\nFull-period Pearson r (Gold vs USD/INR): {r_full:+.4f}  "
      f"(p = {p_full:.4f}, n = {len(gold_ret.dropna())})")

# ── Scatter plot with regression line ─────────────────────────────────────────
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Left: scatter + regression line coloured by year
ax = axes[0]
years  = gold_ret.index.year
cmap   = plt.cm.get_cmap('plasma', years.max() - years.min() + 1)
colors = [cmap(y - years.min()) for y in years]
scatter = ax.scatter(gold_ret.values, usdinr_ret.values,
                     c=years, cmap='plasma', alpha=0.7, s=35, zorder=3)
# OLS regression line
m, c_val, _, _, _ = stats.linregress(gold_ret.values, usdinr_ret.values)
x_line = np.linspace(gold_ret.min(), gold_ret.max(), 100)
ax.plot(x_line, m * x_line + c_val, color='crimson', linewidth=2,
        label=f'OLS fit  r={r_full:+.3f}')
ax.axhline(0, color='gray', linewidth=0.7)
ax.axvline(0, color='gray', linewidth=0.7)
ax.set_xlabel('Gold Monthly Return', fontsize=11)
ax.set_ylabel('USD/INR Monthly Change', fontsize=11)
ax.set_title('Gold Returns vs USD/INR Change — Scatter (2010–2025)',
             fontsize=11, fontweight='bold')
plt.colorbar(scatter, ax=ax, label='Year')
ax.legend(fontsize=9)

# Right: 12-month rolling correlation between Gold and USD/INR
ax2 = axes[1]
rolling_gold_usdinr = (
    gold_ret.rolling(12, min_periods=12)
    .corr(usdinr_ret)
)
ax2.plot(rolling_gold_usdinr.index, rolling_gold_usdinr.values,
         color='darkorange', linewidth=2, label='12M Rolling r')
ax2.axhline(0, color='black', linewidth=0.8, linestyle=':')
ax2.axhline(r_full, color='crimson', linewidth=1, linestyle='--',
            label=f'Full-period avg r = {r_full:+.3f}')

# Shade COVID and rate-hike periods
ax2.axvspan('2020-01-01', '2021-12-31', alpha=0.15, color='red',
            label='COVID (2020–21)')
ax2.axvspan('2022-01-01', '2023-12-31', alpha=0.15, color='orange',
            label='Rate Hikes (2022–23)')
ax2.set_ylabel('Pearson Correlation')
ax2.set_xlabel('Date')
ax2.set_title('12M Rolling Correlation: Gold vs USD/INR',
              fontsize=11, fontweight='bold')
ax2.legend(fontsize=8)
ax2.xaxis.set_major_formatter(mdates.DateFormatter('%Y'))
fig.autofmt_xdate()

plt.suptitle('Safe-Haven Analysis: Gold vs USD/INR Monthly Returns',
             fontsize=13, fontweight='bold')
plt.tight_layout()
plt.show()

# ── Regime breakdown for Gold vs USD/INR ─────────────────────────────────────
print("\nGold vs USD/INR — Regime-specific correlations:")
print(f"{'Regime':<30}  {'r':>8}  {'p-value':>10}  {'n':>4}  Interpretation")
print("-" * 80)
for regime_label, (start, end) in REGIMES.items():
    sub_gold  = gold_ret.loc[start:end].dropna()
    sub_usdinr = usdinr_ret.loc[start:end].dropna()
    common_idx = sub_gold.index.intersection(sub_usdinr.index)
    if len(common_idx) > 5:
        r_r, p_r = stats.pearsonr(
            sub_gold.loc[common_idx].values,
            sub_usdinr.loc[common_idx].values
        )
        # Economic interpretation
        if r_r > 0.2:
            interp = "Positive: risk-off → gold↑ & INR weakens"
        elif r_r < -0.2:
            interp = "Negative: gold rises as USD weakens"
        else:
            interp = "Near-zero: mixed / no clear safe-haven signal"
        print(f"  {regime_label.replace(chr(10), ' '):<28}  "
              f"{r_r:>+8.4f}  {p_r:>10.4f}  {len(common_idx):>4}  {interp}")

````

````python
# ─────────────────────────────────────────────────────────────────────────────
````
````python
# SECTION 9: CORRELATION STABILITY ANALYSIS
#
# For each pair, we compute the STANDARD DEVIATION of the rolling correlation
# over time.  High std = unstable, regime-sensitive relationship.
# Low std = persistent, structural relationship.
#
# This answers: "Which correlations are robust vs which shift dramatically?"
# ─────────────────────────────────────────────────────────────────────────────

print("\n" + "=" * 65)
print(" 9. CORRELATION STABILITY (ROLLING STD OF ROLLING CORRELATION)")
print("=" * 65)

stability_rows = []
for a, b in pairs:
    key   = ('12M Rolling', a, b)
    ser   = rolling_corr[key].dropna()
    r_avg = ser.mean()
    r_std = ser.std()
    r_min = ser.min()
    r_max = ser.max()
    stability_rows.append({
        'Pair':            f'{a} vs {b}',
        'Mean r':          round(r_avg, 4),
        'Std r':           round(r_std, 4),
        'Min r':           round(r_min, 4),
        'Max r':           round(r_max, 4),
        'Range (Max-Min)': round(r_max - r_min, 4),
    })

stability_df = pd.DataFrame(stability_rows).sort_values('Std r',
                                                         ascending=False)
print("\nCorrelation stability — sorted by std (most volatile first):")
print(stability_df.to_string(index=False))

# ── Bar chart of std by pair ──────────────────────────────────────────────────
fig, ax = plt.subplots(figsize=(12, 5))
colors  = ['crimson' if s > 0.3 else 'steelblue'
           for s in stability_df['Std r'].values]
bars    = ax.barh(stability_df['Pair'], stability_df['Std r'],
                  color=colors, edgecolor='white', alpha=0.85)
ax.set_xlabel('Std of 12M Rolling Correlation')
ax.set_title('Correlation Stability — Std of 12M Rolling Pearson r\n'
             '(Higher = more volatile / regime-sensitive)',
             fontsize=11, fontweight='bold')
for bar, val in zip(bars, stability_df['Std r']):
    ax.text(bar.get_width() + 0.005, bar.get_y() + bar.get_height() / 2,
            f'{val:.3f}', va='center', fontsize=9)
ax.axvline(0.3, color='crimson', linewidth=1, linestyle='--',
           alpha=0.7, label='Instability threshold (0.30)')
ax.legend(fontsize=9)
plt.tight_layout()
plt.show()
````

````python
# ─────────────────────────────────────────────────────────────────────────────
# SECTION 10: BRENT OIL DYNAMICS — SPECIAL FOCUS
#
# Brent oil has the most volatile correlation with everything else.
# We examine its rolling correlation with USD/INR (oil import cost channel)
# and with S&P500 (risk appetite channel).
#
# Economic logic:
#  • Brent↑ → India import bill rises → INR depreciates → USD/INR↑
#  • Brent is a risk-on asset → correlated positively with equities in
#    growth phases, but negatively during supply-shock recessions.
# ─────────────────────────────────────────────────────────────────────────────

print("\n" + "=" * 65)
print(" 10. BRENT OIL — FOCUS ANALYSIS (USD/INR AND S&P500)")
print("=" * 65)

brent_col  = 'brent_monthly_return'
brent_ret  = ret_plot['Brent']
sp500_ret  = ret_plot['S&P500']
usdinr_ret2 = ret_plot['USD/INR']

rolling_brent_usdinr = brent_ret.rolling(12, min_periods=12).corr(usdinr_ret2)
rolling_brent_sp500  = brent_ret.rolling(12, min_periods=12).corr(sp500_ret)

fig, axes = plt.subplots(1, 2, figsize=(18, 5))

for ax, rolling_ser, pair_label, color in zip(
        axes,
        [rolling_brent_usdinr, rolling_brent_sp500],
        ['Brent vs USD/INR', 'Brent vs S&P500'],
        ['darkorange', 'steelblue']):
    ax.plot(rolling_ser.index, rolling_ser.values, color=color, linewidth=2)
    ax.axhline(0, color='black', linewidth=0.7, linestyle=':')
    ax.axhline(rolling_ser.mean(), color='crimson', linewidth=1,
               linestyle='--', alpha=0.6,
               label=f'Mean = {rolling_ser.mean():+.2f}')
    ax.axvspan('2020-01-01', '2021-12-31', alpha=0.15, color='red',
               label='COVID')
    ax.axvspan('2022-01-01', '2023-12-31', alpha=0.15, color='orange',
               label='Rate Hikes')
    ax.set_title(f'12M Rolling Correlation: {pair_label}',
                 fontsize=11, fontweight='bold')
    ax.set_ylabel('Pearson r')
    ax.set_ylim(-1.05, 1.05)
    ax.legend(fontsize=8)
    ax.xaxis.set_major_formatter(mdates.DateFormatter('%Y'))
    fig.autofmt_xdate()

plt.suptitle('Brent Oil — Cross-Market Correlation Dynamics',
             fontsize=13, fontweight='bold')
plt.tight_layout()
plt.show()

# Numerical summary
r_brent_usdinr, p_b_usdinr = stats.pearsonr(
    brent_ret.dropna().values, usdinr_ret2.loc[brent_ret.dropna().index].values
)
r_brent_sp500, p_b_sp500 = stats.pearsonr(
    brent_ret.dropna().values, sp500_ret.loc[brent_ret.dropna().index].values
)
print(f"\nBrent vs USD/INR  — full-period r = {r_brent_usdinr:+.4f}  "
      f"(p = {p_b_usdinr:.4f})")
print(f"Brent vs S&P500   — full-period r = {r_brent_sp500:+.4f}  "
      f"(p = {p_b_sp500:.4f})")

````

````python
# ─────────────────────────────────────────────────────────────────────────────
# SECTION 11: CROSS-CORRELATION AT LAGS (LEAD-LAG ANALYSIS)
#
# Cross-correlation at lag k tests: does asset A at time t predict asset B
# at time t+k?  If the US leads India, we expect positive cross-correlation
# at lag k=1 (S&P500 at t → Nifty50 at t+1).
#
# This is complementary to the Granger causality tests done in Module 3.
# ─────────────────────────────────────────────────────────────────────────────

print("\n" + "=" * 65)
print(" 11. LEAD-LAG CROSS-CORRELATION (S&P500 LEADS NIFTY50?)")
print("=" * 65)

nifty = ret_plot['Nifty50'].dropna()
sp500 = ret_plot['S&P500'].loc[nifty.index]

MAX_LAGS = 6  # test up to 6 months ahead / behind

lags      = range(-MAX_LAGS, MAX_LAGS + 1)
cross_corr_vals = []

for lag in lags:
    if lag == 0:
        r, p = stats.pearsonr(sp500.values, nifty.values)
    elif lag > 0:
        # S&P500 leads Nifty50 by `lag` months
        r, p = stats.pearsonr(sp500.values[:-lag], nifty.values[lag:])
    else:
        # Nifty50 leads S&P500 by `abs(lag)` months
        r, p = stats.pearsonr(sp500.values[abs(lag):], nifty.values[:lag])
    cross_corr_vals.append({'Lag': lag, 'r': round(r, 4), 'p': round(p, 4)})

cross_corr_df = pd.DataFrame(cross_corr_vals)
print("\nCross-correlation: S&P500(t) vs Nifty50(t+lag)")
print("Positive lag = S&P500 leads Nifty50, Negative lag = Nifty50 leads S&P500")
print(cross_corr_df.to_string(index=False))

# ── Bar chart ─────────────────────────────────────────────────────────────────
fig, ax = plt.subplots(figsize=(12, 5))
colors  = ['steelblue' if r > 0 else 'crimson'
           for r in cross_corr_df['r'].values]
ax.bar(cross_corr_df['Lag'], cross_corr_df['r'],
       color=colors, edgecolor='white', alpha=0.85)
ax.axhline(0, color='black', linewidth=0.8)
ax.set_xlabel('Lag (months)  — positive = S&P500 leads Nifty50',
              fontsize=10)
ax.set_ylabel('Pearson r')
ax.set_title('Lead-Lag Cross-Correlation: S&P500 → Nifty50\n'
             '(Positive lag = US leads India)',
             fontsize=11, fontweight='bold')
ax.set_xticks(list(lags))
plt.tight_layout()
plt.show()

# Best lead-lag
best_row = cross_corr_df.loc[cross_corr_df['r'].abs().idxmax()]
print(f"\nStrongest cross-correlation at lag {int(best_row['Lag'])}: "
      f"r = {best_row['r']:+.4f}  (p = {best_row['p']:.4f})")
````
